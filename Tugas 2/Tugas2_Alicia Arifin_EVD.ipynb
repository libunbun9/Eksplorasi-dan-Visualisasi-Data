## Penanganan Missing Data
Penanganan Missing pada data

### Missing Completely at Random
Data hilang secara acak, dan tidak terkait dengan variabel lain dalam dataset. Untuk mengatasinya bisa melakukan imputasi atau penghapusan baris/kolom.

### Missing at Random
Data hilang terkait dengan satu atau lebih variabel lainnya, tetapi tidak dengan variabel yang hilang itu sendiri. Untuk mengatasinya bisa melakukan imputasi atau metode canggih seperti multiple imputation

### Missing Not at Random
Data hilang tidak secara acak, dan hilangnya data terkait langsung dengan variabel yang hilang itu sendiri. Untuk mengatasinya bisa melakukan tahap yang lebih kompleks, bisa menggunakan model statistik untuk memperkirakan data yang hilang.
